**Bài 1:**
Một loại sản phẩm X được bán ra thị trường do một nhà máy gồm ba phân xưởng I, II và
III sản xuất, trong đó phân xưởng I chiếm 35%, phân xưởng II chiếm 40% và phân xưởng
III chiếm 25%. Tỷ lệ sản phẩm loại A do ba phân xưởng I, II và III sản xuất lần lượt là
80%, 60% và 90%.  
1) Tính tỷ lệ sản phẩm loại A nói chung do nhà máy sản xuất.  
2) Chọn mua ngẫu nhiên một sản phẩm X ở thị trường. Giả sử đã mua được sản phẩm loại  
A. Khi đó theo bạn, sản phẩm được mua do phân xưởng nào sản xuất là có khả năng nhất?  
3) Chọn mua ngẫu nhiên 10 sản phẩm X ở thị trường. Tính xác suất để có đúng 7 sản phẩm  
loại A. 

```python
#1.1
p1, p2, p3 = 0.35,0.4,0.25
tyleA1, tyleA2, tyleA3 = 0.8, 0.6,0.9
p = p1*tyleA1 + p2*tyleA2 + p3*tyleA3
print("Tỷ lệ sản phảm loại A là: ", p)
```

```python
#1.2
pA1 = (tyleA1*p1)/p
pA2 = (tyleA2*p2)/p
pA3 = (tyleA3*p3)/p
phanxuong = {"Phân xưởng I": pA1, "Phân xưởng II": pA2, "Phân xưởng III": pA3}
most_likely = max(phanxuong, key=phanxuong.get)
print(f"2) Xác suất hậu nghiệm: I={pA1:.4f}, II={pA2:.4f}, III={pA3:.4f}")
print(f"   => Khả năng nhất là: {most_likely}")
```

```python
#1.3
prob_7_A = binom.pmf(7,10,p)
print(f" Xác suất có đúng 7 sản phẩm loại A trong 10 sản phẩm: {prob_7_A:.4f}\n")
```

**Bài 4:**
Tỷ lệ người dân tham gia trên mạng lưới giao dịch có sự hiểu biết về luật giao dịch là 90%.
Tại một nút giao diện có 5 người giao diện trong phạm vi luật. Gọi X là một số người am
hiểu luật nhưng cố gắng tình luật trong 5 người đó.  
a) Lập bảng phân phối của X.  
b) Tính E (3- 5X) và V (3X-2).  
c) Nếu có 30 người trong phạm vi luật, thì trung bình có bao nhiêu người hiểu biết luật
nhưng cố gắng trong phạm vi? Có bao nhiêu người trong phạm vi luật làm người không
hiểu luật?


```python
#4.a) X ~ B(5, 0.9)
n_4, p_4 = 5, 0.9
```

```python
#4.b
ex = n_4 * p_4
vx = n_4 * p_4 * (1 - p_4)
e_transformed = 3 - 5 * ex
v_transformed = (3**2) * vx
print(f" E(3-5X) = {e_transformed}, V(3X-2) = {v_transformed}")
```

```python
#4.c 
n_30 = 30
avg_know = n_30 * 0.9
avg_not_know = n_30 * 0.1
print(f" Trung bình số người hiểu biết luật (n=30): {avg_know}")
print(f" Trung bình số người không hiểu biết luật: {avg_not_know}\n")
```

```python
#5.a) X có phân phối Nhị thức B(5, 0.2)
p_2 = 2/10
```

**Bài 6:**
Chiều dài cá được mô hình hóa bằng phân phối chuẩn N(=16 (cm), =4 (cm)). Thực hiện
các yêu cầu sau:  
- Xác suất bắt được con cá nhỏ (nhỏ hơn 6 (cm))?  
- Giả sử, ai bắt được con cá lớn (lớn hơn 20(cm)) sẽ được thưởng. Hỏi xác suất được
thưởng là bao nhiêu?  
- Xác suất bắt được con cá vừa (trong khoảng 10-24(cm))?  

```python
mu, sigma = 16, 4
# Cá nhỏ (<6cm)
p_small = norm.cdf(6, mu, sigma)
# Cá lớn (>20cm)
p_large = 1 - norm.cdf(20, mu, sigma)
# Cá vừa (10-24cm)
p_medium = norm.cdf(24, mu, sigma) - norm.cdf(10, mu, sigma)

print(f"Xác suất bắt được cá nhỏ (<6cm): {p_small:.4f}")
print(f"Xác suất được thưởng (>20cm): {p_large:.4f}")
print(f"Xác suất bắt được cá vừa (10-24cm): {p_medium:.4f}")
```

**Bài 7:**
Một giám đốc nhân sự mong muốn so sánh tính hiệu quả của hai phương pháp huấn luyện
các nhân viên công nghiệp nhằm thực hiện một hoạt động lắp ráp nào đó. Một số lượng
nhân viên được chia thành hai nhóm bằng nhau, n hóm thứ nhất nhận được phương pháp
huấn luyện 1 và nhóm thứ hai được huấn luyện bằng phương pháp 2. Mỗi nhóm sẽ thực
hiện hoạt động lắp ráp này, và độ dài của thời gian lắp ráp sẽ được ghi nhận. Người ta kỳ
vọng rằng các đại lượng cho cả hai nhóm sẽ có một khoảng xấp xỉ 8 phút. Để cho ước
lượng về sự khác biệt về thời gian trung bình để lắp ráp chính xác trong giới hạn 1 phút với
xác suất bằng với 0.95, thì cần phải đưa bao nhiêu công nhân vào mỗi nhóm huấn luyện?

```python
# E = 1, confidence = 0.95 => z = 1.96. Giả sử độ lệch chuẩn sigma_train xấp xỉ 2 (từ range 8)
z = 1.96
E = 1
sigma_train = 2 
n_sample = (z**2 * sigma_train**2) / (E**2)
print(f"Số công nhân cần thiết mỗi nhóm (ước lượng): {math.ceil(n_sample)}")
```

```python
#LAB 1 – Multi-Source Market Data Ingestion
csv_df = pd.read_csv("prices.csv")

# Load JSON
json_df = pd.read_json("alt_data.json")

def standardize(df):
    df['timestamp'] = pd.to_datetime(df['timestamp'], utc=True)
    return df[['timestamp','symbol','open','high','low','close','volume']]

csv_df = standardize(csv_df)
json_df = standardize(json_df)

# Merge
market_df = pd.concat([csv_df, json_df]).drop_duplicates()

print(market_df.head())
```

```python
#LAB 2 – Incremental Market Data Loading (ĐÃ SỬA)
import json
import os

META_FILE = "metadata.json"

# Load metadata
if os.path.exists(META_FILE):
    with open(META_FILE) as f:
        meta = json.load(f)
else:
    meta = {}

incremental = []

for sym, g in market_df.groupby("symbol"):
    last_ts = meta.get(sym)

    if last_ts:
        g = g[g['timestamp'] > pd.to_datetime(last_ts)]

    g = g.drop_duplicates(subset=['timestamp'])

    if not g.empty:
        meta[sym] = g['timestamp'].max().isoformat()
        incremental.append(g)

# Save metadata
with open(META_FILE, "w") as f:
    json.dump(meta, f, indent=2)

inc_df = pd.concat(incremental) if incremental else pd.DataFrame()
print(inc_df.shape)
```

```python
#LAB 3 – Corporate Actions Adjustment (ĐÃ SỬA)
prices = pd.read_csv("prices.csv")
prices['timestamp'] = pd.to_datetime(prices['timestamp'], utc=True)

actions = pd.read_csv("corp_actions.csv")
actions['date'] = pd.to_datetime(actions['date'], utc=True)

# Apply split (backward)
for _, r in actions.dropna(subset=['split']).iterrows():
    mask = (prices.symbol == r.symbol) & (prices.timestamp < r.date)
    prices.loc[mask, ['open','high','low','close']] /= r.split

# Apply dividend
for _, r in actions.dropna(subset=['dividend']).iterrows():
    mask = (prices.symbol == r.symbol) & (prices.timestamp < r.date)
    prices.loc[mask, 'close'] -= r.dividend

# Validate returns
prices['return'] = prices.groupby('symbol')['close'].pct_change()
print(prices[['timestamp','close','return']].head())
```

```python
#LAB 4 – FX Tick Data Resampling (CSV)
ticks = pd.read_csv("fx_ticks.csv")
ticks['timestamp'] = pd.to_datetime(ticks['timestamp'], utc=True)

ticks = ticks.sort_values('timestamp').drop_duplicates()

# Resample 1s
ohlc_1s = ticks.set_index('timestamp')['price'].resample('1s').ohlc()

# Resample 1m
ohlc_1m = ticks.set_index('timestamp')['price'].resample('1min').ohlc()

print(ohlc_1s.head())
```

```python
#LAB 5 – Futures Continuous Contract
futures = pd.read_csv("futures.csv")
futures['date'] = pd.to_datetime(futures['date'], utc=True)

# chọn contract có volume lớn nhất mỗi ngày
futures['rank'] = futures.groupby('date')['volume'].rank(ascending=False)
continuous = futures[futures['rank'] == 1].sort_values('date')

continuous['return'] = continuous['close'].pct_change()
print(continuous.head())
```

```python
#LAB 6 – Feature Engineering
import numpy as np

df = prices.copy()

df['log_return'] = np.log(df['close']).diff()
df['vol_20'] = df['log_return'].rolling(20).std()
df['momentum_5'] = df['close'] / df['close'].shift(5) - 1

print(df[['log_return','vol_20','momentum_5']].head())
```

```python
#LAB 7 – Cross-Asset Correlation
wide = df.pivot(index='timestamp', columns='symbol', values='log_return')
rolling_corr = wide.rolling(20).corr()

print(rolling_corr.tail())
```

```python
#LAB 8 – Market Microstructure Metrics
ticks['spread'] = ticks['ask'] - ticks['bid']
ticks['imbalance'] = (ticks['bid_size'] - ticks['ask_size']) / \
                     (ticks['bid_size'] + ticks['ask_size'])

print(ticks[['spread','imbalance']].head())
```

```python
#LAB 9 – Intraday Session Aggregation
df['hour'] = df['timestamp'].dt.hour

df['session'] = pd.cut(
    df['hour'],
    bins=[9, 11, 14, 16],
    labels=['Morning','Midday','Afternoon']
)

session_agg = df.groupby('session').agg({
    'volume':'sum',
    'log_return':'mean'
})

print(session_agg)
```

```python
#LAB 10 – Survivorship Bias
current = pd.read_csv("current_universe.csv")

hist_symbols = set(df.symbol.unique())
current_symbols = set(current.symbol)

biased = hist_symbols - current_symbols
print("Symbols gây survivorship bias:", biased)
```
